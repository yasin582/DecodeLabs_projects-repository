```python
# Advanced EDA and Feature Engineering

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

pd.set_option('display.max_columns', None)

# Loading the data

df = pd.read_excel('/content/Dataset for Data Analytics.xlsx')

print("Shape of dataset:", df.shape)
df.head()


# Basic information about the dataset

df.info()

print("\nSummary of data:")
display(df.describe(include='all').T)


# Checking unique values

for col in df.columns:
    print(col, ":", df[col].nunique())


# Checking missing values

print("\nMissing values:")
display(df.isnull().sum())


# Percentage of missing values

missing_percent = (df.isnull().sum() / len(df)) * 100
display(missing_percent[missing_percent > 0])


# Checking duplicate rows

print("Number of duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)


# Converting columns into the correct data types

# Rename 'Date' column to 'OrderDate' for consistency with code
df = df.rename(columns={'Date': 'OrderDate'})
df['OrderDate'] = pd.to_datetime(df['OrderDate'], errors='coerce')

df['Quantity'] = pd.to_numeric(
    df['Quantity'],
    errors='coerce'
)

df['UnitPrice'] = pd.to_numeric(
    df['UnitPrice'],
    errors='coerce'
)


# Removing rows with missing important values

df = df.dropna(
    subset=[
        'CustomerID',
        'OrderDate',
        'Quantity',
        'UnitPrice'
    ]
)

# Removing invalid quantity and price values

df = df[
    (df['Quantity'] > 0) &
    (df['UnitPrice'] > 0)
].copy()

print("Shape after cleaning:", df.shape)


# CouponCode has some missing values

df['CouponCode'] = df['CouponCode'].fillna('No Coupon')

print("\nCoupon information:")
display(df['CouponCode'].value_counts().head(10))


# ---------------------------------------
# Exploratory Data Analysis
# ---------------------------------------

# Quantity distribution

plt.figure(figsize=(8,5))
sns.histplot(df['Quantity'], bins=50, kde=True)

plt.title('Distribution of Quantity')
plt.xlabel('Quantity')
plt.ylabel('Number of Orders')

plt.show()


# Unit price distribution

plt.figure(figsize=(8,5))
sns.histplot(df['UnitPrice'], bins=50, kde=True)

plt.title('Distribution of Unit Price')
plt.xlabel('Unit Price')
plt.ylabel('Number of Orders')

plt.show()


# ---------------------------------------
# Feature Engineering
# ---------------------------------------

# Total price for each transaction

df['TotalPrice'] = (
    df['Quantity'] *
    df['UnitPrice']
)

# Average price per item

df['AvgItemCost'] = (
    df['TotalPrice'] /
    df['Quantity']
)

# Checking whether a coupon was used
# 1 means coupon was used

df['IsDiscounted'] = (
    df['CouponCode'] != 'No Coupon'
).astype(int)


# Creating date features

df['OrderYear'] = df['OrderDate'].dt.year
df['OrderMonth'] = df['OrderDate'].dt.month
df['OrderDay'] = df['OrderDate'].dt.day

df['OrderDayOfWeek'] = (
    df['OrderDate'].dt.day_name()
)

# Weekend or not

df['IsWeekend'] = (
    df['OrderDate'].dt.dayofweek >= 5
).astype(int)

display(df.head())


# ---------------------------------------
# Sales Analysis
# ---------------------------------------

print("Total sales:",
      round(df['TotalPrice'].sum(), 2))


# Sales by year

yearly_sales = (
    df.groupby('OrderYear')['TotalPrice']
      .sum()
)

display(yearly_sales)


plt.figure(figsize=(8,5))

sns.barplot(
    x=yearly_sales.index,
    y=yearly_sales.values
)

plt.title('Sales by Year')
plt.xlabel('Year')
plt.ylabel('Total Sales')

plt.show()


# Monthly sales

monthly_sales = (
    df.groupby(
        df['OrderDate'].dt.to_period('M')
    )['TotalPrice']
    .sum()
)

plt.figure(figsize=(14,5))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker='o'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')

plt.xticks(rotation=45)

plt.show()


# Orders by day of week

days = [
    'Monday',
    'Tuesday',
    'Wednesday',
    'Thursday',
    'Friday',
    'Saturday',
    'Sunday'
]

day_orders = (
    df['OrderDayOfWeek']
    .value_counts()
    .reindex(days)
)

plt.figure(figsize=(9,5))

sns.barplot(
    x=day_orders.index,
    y=day_orders.values
)

plt.title('Orders by Day of Week')
plt.xlabel('Day')
plt.ylabel('Number of Orders')

plt.xticks(rotation=30)

plt.show()


# Coupon usage

print("Orders without coupon:",
      (df['IsDiscounted'] == 0).sum())

print("Orders with coupon:",
      (df['IsDiscounted'] == 1).sum())


plt.figure(figsize=(6,5))

sns.countplot(
    data=df,
    x='IsDiscounted'
)

plt.title('Coupon Usage')
plt.xlabel('Coupon Used (0 = No, 1 = Yes)')
plt.ylabel('Number of Orders')

plt.show()


# ---------------------------------------
# Outlier Analysis
# ---------------------------------------

columns = [
    'Quantity',
    'UnitPrice',
    'TotalPrice'
]

for col in columns:

    plt.figure(figsize=(8,4))

    sns.boxplot(
        x=df[col]
    )

    plt.title('Boxplot of ' + col)
    plt.xlabel(col)

    plt.show()


# Finding the number of outliers

for col in columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    print(
        col,
        "outliers:",
        len(outliers)
    )


# Capping very high values instead of deleting them

df['Quantity_Capped'] = df['Quantity'].clip(
    upper=df['Quantity'].quantile(0.99)
)

df['UnitPrice_Capped'] = df['UnitPrice'].clip(
    upper=df['UnitPrice'].quantile(0.99)
)

df['TotalPrice_Capped'] = (
    df['Quantity_Capped'] *
    df['UnitPrice_Capped']
)


# ---------------------------------------
# Correlation
# ---------------------------------------

corr_columns = [
    'Quantity',
    'UnitPrice',
    'TotalPrice',
    'AvgItemCost',
    'IsDiscounted',
    'IsWeekend'
]

corr = df[corr_columns].corr()

plt.figure(figsize=(9,6))

sns.heatmap(
    corr,
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)

plt.title('Correlation Matrix')

plt.show()


# ---------------------------------------
# RFM Analysis
# ---------------------------------------

# R = how recently the customer purchased
# F = how many times the customer purchased
# M = how much the customer spent

reference_date = (
    df['OrderDate'].max()
    + pd.Timedelta(days=1)
)

rfm = df.groupby('CustomerID').agg(
    Recency=(
        'OrderDate',
        lambda x:
        (reference_date - x.max()).days
    ),

    Frequency=(
        'OrderID',
        'nunique'
    ),

    Monetary=(
        'TotalPrice',
        'sum'
    )
).reset_index()

print("Number of customers:", len(rfm))

display(rfm.head())


# RFM summary

display(
    rfm[
        ['Recency', 'Frequency', 'Monetary']
    ].describe()
)


# RFM distributions

for col in [
    'Recency',
    'Frequency',
    'Monetary'
]:

    plt.figure(figsize=(8,5))

    sns.histplot(
        rfm[col],
        kde=True
    )

    plt.title(col + ' Distribution')
    plt.xlabel(col)
    plt.ylabel('Customers')

    plt.show()


# ---------------------------------------
# Preparing RFM for clustering
# ---------------------------------------

# The values are quite skewed, so log transformation
# makes the distributions a little easier to work with.

rfm_model = rfm.copy()

rfm_model['Recency'] = np.log1p(
    rfm_model['Recency']
)

rfm_model['Frequency'] = np.log1p(
    rfm_model['Frequency']
)

rfm_model['Monetary'] = np.log1p(
    rfm_model['Monetary']
)


# Scaling the values

scaler = StandardScaler()

rfm_scaled = scaler.fit_transform(
    rfm_model[
        ['Recency', 'Frequency', 'Monetary']
    ]
)


# ---------------------------------------
# Finding a suitable number of clusters
# ---------------------------------------

inertia = []
silhouette_scores = []

for k in range(2, 9):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(rfm_scaled)

    inertia.append(model.inertia_)

    silhouette_scores.append(
        silhouette_score(
            rfm_scaled,
            labels
        )
    )


# Elbow method

plt.figure(figsize=(8,5))

plt.plot(
    range(2,9),
    inertia,
    marker='o'
)

plt.title('Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')

plt.show()


# Silhouette score

plt.figure(figsize=(8,5))

plt.plot(
    range(2,9),
    silhouette_scores,
    marker='o'
)

plt.title('Silhouette Score')
plt.xlabel('Number of Clusters')
plt.ylabel('Score')

plt.show()


# Display the scores

scores = pd.DataFrame({
    'Clusters': range(2,9),
    'Silhouette Score': silhouette_scores
})

display(scores)


# ---------------------------------------
# K-Means
# ---------------------------------------

# I am using 4 clusters for the final segmentation.

kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

rfm['Cluster'] = kmeans.fit_predict(
    rfm_scaled
)

display(rfm.head())


# ---------------------------------------
# Cluster analysis
# ---------------------------------------

cluster_summary = rfm.groupby(
    'Cluster'
).agg(
    Customers=('CustomerID', 'count'),
    Avg_Recency=('Recency', 'mean'),
    Avg_Frequency=('Frequency', 'mean'),
    Avg_Monetary=('Monetary', 'mean')
)

cluster_summary['Customer_Percentage'] = (
    cluster_summary['Customers'] /
    cluster_summary['Customers'].sum() * 100
)

display(cluster_summary)


# ---------------------------------------
# Customer segmentation plot
# ---------------------------------------

plt.figure(figsize=(10,6))

sns.scatterplot(
    data=rfm,
    x='Recency',
    y='Monetary',
    hue='Cluster',
    size='Frequency',
    sizes=(40,250),
    alpha=0.7
)

plt.title('Customer Segmentation')
plt.xlabel('Recency')
plt.ylabel('Monetary Value')

plt.show()


# ---------------------------------------
# Checking each cluster
# ---------------------------------------

for cluster in sorted(rfm['Cluster'].unique()):

    print("\nCluster:", cluster)

    cluster_data = rfm[
        rfm['Cluster'] == cluster
    ]

    display(
        cluster_data[
            ['Recency',
             'Frequency',
             'Monetary']
        ].describe()
    )


# ---------------------------------------
# Give meaningful names to the clusters
# ---------------------------------------

# Check the cluster_summary above and change
# the names if the characteristics are different.

cluster_names = {
    0: 'Regular Customers',
    1: 'At-Risk Customers',
    2: 'Loyal Customers',
    3: 'High Value Customers'
}

rfm['Segment'] = rfm['Cluster'].map(
    cluster_names
)


# Final customer data

final_data = rfm[
    [
        'CustomerID',
        'Recency',
        'Frequency',
        'Monetary',
        'Cluster',
        'Segment'
    ]
].sort_values(
    'Monetary',
    ascending=False
)

display(final_data.head(20))


# ---------------------------------------
# Final segment summary
# ---------------------------------------

segment_summary = rfm.groupby(
    'Segment'
).agg(
    Customers=('CustomerID', 'count'),
    Avg_Recency=('Recency', 'mean'),
    Avg_Frequency=('Frequency', 'mean'),
    Avg_Monetary=('Monetary', 'mean')
).sort_values(
    'Avg_Monetary',
    ascending=False
)

segment_summary['Customer_Share_%'] = (
    segment_summary['Customers'] /
    segment_summary['Customers'].sum() * 100
)

display(segment_summary)


# Number of customers in each segment

plt.figure(figsize=(9,5))

sns.barplot(
    data=segment_summary.reset_index(),
    x='Segment',
    y='Customers'
)

plt.title('Customers in Each Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Number of Customers')

plt.xticks(rotation=30)

plt.show()


# Revenue contribution by segment

revenue = (
    rfm.groupby('Segment')['Monetary']
       .sum()
       .sort_values(ascending=False)
)

plt.figure(figsize=(9,5))

sns.barplot(
    x=revenue.index,
    y=revenue.values
)

plt.title('Revenue Contribution by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Total Revenue')

plt.xticks(rotation=30)

plt.show()


# ---------------------------------------
# Final results
# ---------------------------------------

print("Final customer segmentation:")

display(
    final_data.head(20)
)


# Saving the results

final_data.to_csv(
    '/content/customer_segments.csv',
    index=False
)

segment_summary.to_csv(
    '/content/customer_segment_summary.csv'
)

print("Analysis completed successfully.")
```
